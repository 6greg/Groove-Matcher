# Train CCA

## Parametri

```python
from pathlib import Path
import sys
import joblib

project_root = Path.cwd().parent
sys.path.insert(0, str(project_root))
from source.cca import load_tensor, fit

tensors_dir = project_root / "data" / "results" / "tensors"
models_dir = project_root / "data" / "model"

arrangement_types = ["bass", "full", "full_no_bass", "harm", "lead"]
```

## CCA 

```python
for arrangement_type in arrangement_types:
    tensor = load_tensor(tensors_dir / f"tensor_train_{arrangement_type}.npz")
    model = fit(tensor["X"], tensor["Y"])
    joblib.dump(model, models_dir / f"cca_{arrangement_type}.joblib")
```
