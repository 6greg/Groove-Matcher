# Train CCA

## Parametri

```python
from pathlib import Path
import sys
import joblib

project_root = Path.cwd().parent
sys.path.insert(0, str(project_root))
from source.cca import load_tensor, fit

tensors_dir = project_root / "data" / "results" / "tensors"
models_dir = project_root / "data" / "model"

arrangement_types = ["bass", "full", "full_no_bass", "harm", "lead"]
```

## CCA 

```python
for arrangement_type in arrangement_types:
    tensor = load_tensor(tensors_dir / f"tensor_train_{arrangement_type}.npz")
    model = fit(tensor["X"], tensor["Y"])
    joblib.dump(model, models_dir / f"cca_{arrangement_type}.joblib")
```

# Interpretabilità modello

```python
for arrangement_type in arrangement_types:
    cca = joblib.load(models_dir / f"cca_{arrangement_type}.joblib")["cca"]
    print(arrangement_type)
    print("pesi a1: ", cca.x_weights_[:, 0].round(2))
    print("pesi b1: ", cca.y_weights_[:, 0].round(2))
    print("arrangiamento: ", cca.x_loadings_[:, 0].round(2))
    print("batteria: ", cca.y_loadings_[:, 0].round(2))
```
